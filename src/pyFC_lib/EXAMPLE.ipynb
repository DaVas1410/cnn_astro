See the contents of `cubegen.py` for specific details on the use of `pyFC`.

```python
# This class uses pyFC's LogNormalFractalCube class
from cubegen import CubeGenerator

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Let's create an instance with the INI parameters file
cloud = CubeGenerator("cloud-params.ini")
```

```python
# And generate the fractal cloud data
fractal_data = cloud.getData()
```

```python
plt.figure(figsize=(5,5), dpi = 100)
plt.title("Fractal cube slice")
plt.imshow(np.log10(fractal_data[:, :, 0]), cmap = 'bwr')
plt.show()
```
